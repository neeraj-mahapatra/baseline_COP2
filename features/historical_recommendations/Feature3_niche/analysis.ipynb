```python
import pandas as pd 
import numpy as np
```

```python
file_path = "data/product_id vs catgory.csv"
df = pd.read_csv(file_path)
df.head
```

```python
df['CODPRODUCTOSAP'].nunique()
```

```python
def hierarchy_to_dict(hierarchy):
    if not isinstance(hierarchy, str):  # Check if the value is a string
        return None  # Return None for NaN or non-string values
    parts = hierarchy.split('/')
    keys = ['Category', 'Sub_Category', 'Type', 'Sub_Type']  # Adjust keys based on levels
    return {key: parts[i] if i < len(parts) else None for i, key in enumerate(keys)}

# Convert the product_hierarchy column to a nested dictionary
df['nested_hierarchy'] = df['DESSUBTIPO'].apply(hierarchy_to_dict)

# Create final dictionary with product_id as keys
result_dict = df.set_index('CODPRODUCTOSAP')['nested_hierarchy'].to_dict()

```

```python
import json

# Save the dictionary to a JSON file
with open('product_hierarchy.json', 'w') as json_file:
    json.dump(result_dict, json_file)
```

```python
c_file_path = "data/product_id vs consultora_id vs date.csv"
c_df = pd.read_csv(c_file_path)
c_df.head
```

```python
c_df['FECHAPROCESO'] = pd.to_datetime(c_df['FECHAPROCESO'])
```

```python
user_product_dict = c_df.groupby('CODEBELISTA').apply(
    lambda x: [(row['CODPRODUCTOSAP'], row['FECHAPROCESO'].strftime('%Y-%m-%d')) for _, row in x.iterrows()]
).to_dict()

# Save the resulting dictionary to a JSON file
with open('user_product_data.json', 'w') as json_file:
    json.dump(user_product_dict, json_file, indent=4)

print("Data has been saved to 'user_product_data.json'")
```
